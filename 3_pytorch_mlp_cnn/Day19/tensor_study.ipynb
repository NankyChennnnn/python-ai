### 张量的类型转换

```python

# 创建一个整型张量
int_tensor = torch.tensor([1, 2, 3, 4], dtype=torch.int32)
print("int32 张量:", int_tensor, int_tensor.dtype)

# 整型张量转换为浮点型张量
float_tensor = int_tensor.float()
print("转换为 float32:", float_tensor, float_tensor.dtype)

# 浮点型张量转换为整型张量
float_tensor2 = torch.tensor([1.8, 2.6, 3.0, 4.9])
int_tensor2 = float_tensor2.int()
print("float32 张量:", float_tensor2, float_tensor2.dtype)
print("转换为 int32 后:", int_tensor2, int_tensor2.dtype)

# 转换为不同位宽的浮点数
float16_tensor = float_tensor2.to(torch.float16)
float64_tensor = float_tensor2.to(torch.float64)
print("float16:", float16_tensor, float16_tensor.dtype)
print("float64:", float64_tensor, float64_tensor.dtype)

# Bool类型张量
bool_tensor = torch.tensor([1, 0, 3, 0]).bool()
print("bool 张量:", bool_tensor, bool_tensor.dtype)

# bool类型转为int类型
int_from_bool = bool_tensor.int()
print("bool 转 int:", int_from_bool, int_from_bool.dtype)

# 使用 .type() 或 .to() 进行类型转换 
converted_by_type = int_tensor.type(torch.float64)
print("通过 .type() 转换为 float64:", converted_by_type, converted_by_type.dtype)

converted_by_to = int_tensor.to(torch.float16)
print("通过 .to() 转换为 float16:", converted_by_to, converted_by_to.dtype)

```

### 矩阵乘法

```python
import torch

# 生成两个整数张量
a = torch.randint(0, 10, (2, 3))
b = torch.randint(0, 10, (3, 2))

print("a:\n", a)
print("b:\n", b)

# 1. 使用 @ 运算符（Python 3.5+ 支持）
result1 = a @ b
print("a @ b 矩阵乘法结果：\n", result1)

# 2. 使用 torch.matmul
result2 = torch.matmul(a, b)
print("torch.matmul(a, b) 结果：\n", result2)

# 3. 使用 torch.mm（仅限二维）
result3 = torch.mm(a, b)
print("torch.mm(a, b) 结果：\n", result3)

# 4. 使用 torch.bmm (必须输入三维batch，举例)
# 扩展成batch形式
a_batch = a.unsqueeze(0)   # [1, 2, 3]
b_batch = b.unsqueeze(0)   # [1, 3, 2]
result4 = torch.bmm(a_batch, b_batch)
print("torch.bmm(a_batch, b_batch) 结果：\n", result4)

# use in transformer
# 5. 使用 torch.einsum
result5 = torch.einsum('ik,kj->ij', a, b)
print("torch.einsum('ik,kj->ij', a, b) 结果：\n", result5)
```

### 形状操作

```python
import torch

# 创建一个 2x3x4 的张量
x = torch.arange(24).reshape(2, 3, 4)
print("原始张量 shape:", x.shape)
print(x)

# 使用 permute 改变维度顺序 (比如转为 4x2x3)
x_permuted = x.permute(2, 0, 1)
print("permute 后的张量 shape:", x_permuted.shape)
print(x_permuted)
```

```python
# squeeze 和 unsqueeze 案例

import torch

# 创建一个带有多余维度的张量
x = torch.randn(1, 3, 1, 5)
print("原始张量 shape:", x.shape)
print(x)

# 使用 squeeze 去除所有为1的维度
x_squeezed = torch.squeeze(x)
print("squeeze 后的张量 shape:", x_squeezed.shape)
print(x_squeezed)

# 只去除某个特定的1维（比如第0维）
x_squeezed0 = torch.squeeze(x, dim=0)
print("只去除第0维后的 shape:", x_squeezed0.shape)

# 使用 unsqueeze 增加一个维度
x_unsqueezed = torch.unsqueeze(x_squeezed, dim=0)
print("在第0维增加一个新维度后的 shape:", x_unsqueezed.shape)

# 还可以在其他位置增加维度
x_unsqueezed2 = torch.unsqueeze(x_squeezed, dim=2)
print("在第2维增加新维度后的 shape:", x_unsqueezed2.shape)
```

### torch中求梯度

```python
# 1. 准备数据：参数与目标值
x = torch.tensor(5.)  # 特征
y = torch.tensor(0.)  # 目标

# 2. 初始化参数：权重与偏置
w = torch.tensor(1., requires_grad=True, dtype=torch.float32)
b = torch.tensor(3., requires_grad=True, dtype=torch.float32)

# 3. 构建方程和损失（均方误差）
z = w * x + b  # 方程 z = w*x + b
loss = (z - y) ** 2  # 均方误差

print("z:", z.item())
print("loss:", loss.item())

# 4. 反向传播，求梯度
loss.backward()

print("w的梯度:", w.grad.item())
print("b的梯度:", b.grad.item())
```

```python
# 1. 生成数据
# 100个样本，1个特征，带一点噪声，偏置为1.5，coef=True：返回权重coef和输出值
from sklearn.datasets import make_regression

# coef 就是 w
x, y, coef = make_regression(n_samples=100, n_features=1, noise=1.0, bias=1.5, coef=True, random_state=42)

import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# 转为 torch tensor
x_tensor = torch.from_numpy(x).float()
y_tensor = torch.from_numpy(y).float().view(-1, 1)

# 定义单层Linear模型
model = nn.Linear(1, 1)

# 优化器的作用是根据损失函数对模型参数（比如权重和偏置）进行自动更新，旨在最小化损失函数。它根据每次反向传播计算得到的梯度，以设定的学习率调整参数，从而逐步优化模型，使其预测结果更加准确。在本例中，我们使用的是SGD（随机梯度下降）优化器。
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.05)

# 训练
epochs = 200  # 训练轮数
losses = []   # 用于记录每轮的loss，后续可用于可视化

for epoch in range(epochs):
    # 前向传播：用当前模型参数对输入数据进行预测
    pred = model(x_tensor)
    # 计算损失：用预测值和真实值计算均方误差损失
    loss = criterion(pred, y_tensor)
    # 记录损失值，便于画loss曲线
    losses.append(loss.item())  # loss.item() 将损失值转化为标量，item() 只能取标量

    # 反向传播前先将梯度清零（否则梯度会累加）
    optimizer.zero_grad()
    # 反向传播：自动计算loss对模型参数的梯度
    loss.backward()
    # 优化器更新模型参数：根据梯度调整权重和偏置  w=w-lr*w.grad
    optimizer.step()

    # 每50轮打印一次当前损失值，便于观察收敛情况
    if (epoch + 1) % 50 == 0:
        print(f"Epoch {epoch+1}/{epochs}, Loss: {loss.item():.4f}")


# 可视化loss曲线
plt.figure(figsize=(6,4))
plt.plot(losses)
plt.title("Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.show()

# 拟合直线可视化
pred_y = model(x_tensor).detach().numpy()

# 拟合前真实直线 y = coef * x + 1.5
true_y = coef * x + 1.5

plt.figure(figsize=(6,4))
plt.scatter(x, y, label='True Data')
plt.plot(x, pred_y, 'r-', label='Fitted Line')
plt.plot(x, true_y, 'g--', label='True Line (coef & bias)')
plt.title("Linear Regression Fit by torch")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()

# 展示权重与偏置
w_trained = model.weight.item()
b_trained = model.bias.item()
print(f"训练后权重: {w_trained:.4f}, 偏置: {b_trained:.4f}")
print(f"真实权重: {coef:.4f}, 真实偏置: 1.5")
```
