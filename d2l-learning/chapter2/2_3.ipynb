### 范数

```python
# L2 范数是向量元素平方和的平方根
u = torch.tensor([3.0, -4.0])
print(torch.norm(u))

# L1 范数表示为向量元素的绝对值之和
print(torch.abs(u).sum())
```

```python
A = torch.arange(24, dtype=torch.float32).reshape(2, 3, 4)
norm_default = torch.linalg.norm(A)
norm_axis0 = torch.linalg.norm(A, dim=0)
norm_axis1 = torch.linalg.norm(A, dim=1)
norm_axis2 = torch.linalg.norm(A, dim=2)
norm_keepdims = torch.linalg.norm(A, dim=(1,2), keepdim=True)
norm_multiple_axes = torch.linalg.norm(A, dim=(0, 2))
(norm_default, norm_axis0.shape, norm_axis1.shape, norm_axis2.shape, norm_keepdims.shape, norm_multiple_axes.shape)
```

```python
# 对任意形状张量进行norm函数测试
B = torch.arange(18, dtype=torch.float32).reshape(3, 2, 3)
# 按默认设置（所有元素范数）
norm_all = torch.linalg.norm(B)
# 沿第0轴计算范数
norm_dim0 = torch.linalg.norm(B, dim=0)
# 沿第1和2轴计算范数，保持维度
norm_dim12 = torch.linalg.norm(B, dim=(1,2), keepdim=True)
# 沿第2轴计算范数，不保持维度
norm_dim2 = torch.linalg.norm(B, dim=2)
(norm_all, norm_dim0, norm_dim0.shape, norm_dim12.shape, norm_dim2.shape)
```
